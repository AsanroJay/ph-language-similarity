## **Cleaning PALITO Dataset**

```python
import re
from pathlib import Path
import unicodedata

def remove_diacritics(text):
    text = unicodedata.normalize("NFD", text)
    text = "".join(
        c for c in text
        if unicodedata.category(c) != "Mn"
    )
    return unicodedata.normalize("NFC", text)

def clean_text(text, patterns):
    text = remove_diacritics(text)

    for pattern in patterns:
        text = re.sub(pattern, "", text)

    text = text.lower()

    text = re.sub(r'["“”]([^"“”]*)["“”]', r'\1', text) #Remove quotation marks
    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r"\n\s*\n+", "\n", text) #Remove \n
    

    return text.strip()
```

### Clean Bikol Text

```python
REGEX_PATTERNS = [
    # Remove titles like "Title: Acts"
    r"([a-zA-Z]+:[A-Z][a-zA-Z]+)",

    # Remove HTML tags like "<b1>Acts</b1>"
    r"([a-zA-Z]+\s+<.+>)",

    # Remove verse numbers surrounded by whitespace
    r"(?<=\s)\d{1,3}(?=\s)",

    # Remove links
    r"https?://\S+",

    # Removes <lbl> content
    r"[A-Za-z]+<lbl>\s*\d+\s*</lbl>",
    r"<lbl>[^<]+</lbl>",

    #Removes other <> tags
    r"</?[^>]+>",

    #Removes punctuations
    r"""[.,;:!?'"()\[\]{}“”‘’\-–—…]""",

    #Removes p\/anything
    r"\b\w+\\/\w+\b"

]
```

```python
input_file = Path("../data/PALITO/Bikol_Religious_Text.txt")
output_file = Path("../data/cleaned/Bikol_Religious_Text.txt")

text = input_file.read_text(encoding="utf-8")

cleaned = clean_text(text, REGEX_PATTERNS)

output_file.parent.mkdir(exist_ok=True)
output_file.write_text(cleaned, encoding="utf-8")

print("Saved religious BIKOL text!")
```

```python
def clean_ceb_text(text, patterns):
    text = remove_diacritics(text)

    for pattern in patterns:
        text = re.sub(pattern, "", text)

    text = text.lower()

    text = re.sub(r'["“”]([^"“”]*)["“”]', r'\1', text) #Remove quotation marks
    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r"\n\s*\n+", "\n", text) #Remove \n
    text = re.sub(r"^\s*-\s*", "", text, flags=re.MULTILINE)
    

    return text.strip()
```

```python
REGEX_PATTERNS = [
    # Remove title lines
    r"(?im)^title\s*:\s*.*$",

    # Remove empty parentheses
    r"\(\s*\)",

    # Remove parenthesized metadata like (mc :; mt :)
    r"\(\s*[a-z]{1,5}\s*[:;,\- ]+[a-z:;,\- ]*\)",

    # Remove <u> tag and content between
    r"<u>\s*.+\s*<pd>\s*.*\s*</pd>\s*.*\s*</u>",

    # Remove HTML/XML tags
    r"</?[^>]+>",

    # Remove URLs
    r"https?://\S+",

    # Remove symbols
    r"[.,;:!?\"'()\[\]{}""¿¡]",

    r"--",

    #Remove - before a sentence
    r"^\s*-\s*"
]
```

```python
input_file = Path("../data/PALITO/Cebuano_Literary_Text.txt")
output_file = Path("../data/cleaned/Cebuano_Literary_Text.txt")

text = input_file.read_text(encoding="utf-8")

cleaned = clean_ceb_text(text, REGEX_PATTERNS)

output_file.parent.mkdir(exist_ok=True)
output_file.write_text(cleaned, encoding="utf-8")


print("Saved literary CEB text!")
```

### **Cleaning Bible Text Collected via API Call**

#### First Combine Individual Chapters into a Consolidated Language Text File

```python
from pathlib import Path

INPUT_DIR = Path("../data/bible_api")

translations = {
    "BSB": "English_Religious_Text.txt",
    "spa_bes": "Spanish_Religious_Text.txt",
}

for prefix, output_name in translations.items():
    files = sorted(INPUT_DIR.glob(f"{prefix}_*.txt"))

    with (INPUT_DIR / output_name).open("w", encoding="utf-8") as outfile:
        for file in files:
            print(f"Adding {file.name}")
            outfile.write(file.read_text(encoding="utf-8"))
            outfile.write("\n\n")

    print(f"Saved {output_name}")
```

```python
input_file = Path("../data/bible_api/English_Religious_Text.txt")
output_file = Path("../data/cleaned/English_Religious_Text.txt")

text = input_file.read_text(encoding="utf-8")

cleaned = clean_text(text, REGEX_PATTERNS)

output_file.parent.mkdir(exist_ok=True)
output_file.write_text(cleaned, encoding="utf-8")

print("Saved English Religious Text!")
```
